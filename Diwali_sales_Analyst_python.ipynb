```python
import pandas as pd
import numpy as np
import seaborn as sns
```

```python
df = pd.read_csv("Diwali Sales Data.csv", encoding= 'unicode_escape')
```

```python
df
```

```python
df.head()
```

```python
df.info()
```

```python
df.drop(["Status", "unnamed1"], axis=1, inplace=True)
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.dropna(inplace=True)
```

```python
df.isnull().sum()
```

```python
df["Amount"] = df["Amount"].astype("int")
```

```python
df.info()
```

```python
df.columns
```

```python
df.rename(columns = {'State':'States'}, inplace=True)
```

```python
df.describe
```

```python
df[['Age', 'Amount']].isnull().sum()
```

```python
df.columns
```

```python
ax = sns.countplot(x = 'Gender', data = df)

for bars in ax.containers:
    ax.bar_label(bars)
```

```python
sales_gen = df.groupby(['Gender'], as_index=False) ['Amount'].sum().sort_values(by='Amount', ascending=False)
```

```python
sns.barplot(x = 'Gender', y='Amount', data = sales_gen)
```

```python
df.columns
```

```python
new = sns.countplot(data = df, x = 'Age Group', hue = 'Gender')

for bars in new.containers:
    new.bar_label(bars)
```

```python
sales_age = df.groupby(['Age Group'], as_index=False) ['Amount'].sum().sort_values(by='Amount', ascending=False)

sns.barplot(x = 'Age Group', y= 'Amount', data = sales_age)
```

```python
df.columns
```

```python
sales_state = df.groupby(['States'], as_index=False) ['Orders'].sum().sort_values(by='Orders', ascending=False).head(10)

sns.set(rc={'figure.figsize':(15,5)})

sns.barplot(data = sales_state, x = 'States', y='Orders')
```

```python
sales_state = df.groupby(['States'], as_index=False) ['Amount'].sum().sort_values(by='Amount', ascending=False).head(10)

sns.set(rc={'figure.figsize':(15,5)})

sns.barplot(data = sales_state, x = 'States', y='Amount')
```

```python
df.columns
```

```python
n = sns.countplot(data = df, x = 'Marital_Status')

sns.set(rc={'figure.figsize':(7,5)})
for bars in n.containers:
    n.bar_label(bars)
```

```python
sales_state = df.groupby(['Marital_Status'], as_index=False) ['Amount'].sum().sort_values(by='Amount', ascending=False)

sns.set(rc={'figure.figsize':(15,5)})

sns.barplot(data = sales_state, x = 'Marital_Status', y='Amount')
```

```python
df.columns
```

```python
sales_state = df.groupby(['Occupation'], as_index=False) ['Amount'].sum().sort_values(by='Amount', ascending=False)

sns.set(rc={'figure.figsize':(20,5)})

sns.barplot(data = sales_state, x = 'Occupation', y='Amount')
```

```python
sns.set(rc={'figure.figsize':(20,5)})

m = sns.countplot(data = df, x = 'Occupation')

for bars in m.containers:
    m.bar_label(bars)
```

```python
sns.set(rc={'figure.figsize':(20,5)})

k = sns.countplot(data = df, x = 'Product_Category')

for bars in k.containers:
    k.bar_label(bars)
```

```python
sales_state = df.groupby(['Product_Category'], as_index=False) ['Amount'].sum().sort_values(by='Amount', ascending=False).head()

sns.set(rc={'figure.figsize':(20,5)})

sns.barplot(data = sales_state, x = 'Product_Category', y='Amount')
```

```python
sales_state = df.groupby(['Product_ID'], as_index=False) ['Amount'].sum().sort_values(by='Amount', ascending=False).head()

sns.set(rc={'figure.figsize':(20,5)})

sns.barplot(data = sales_state, x = 'Product_ID', y='Amount')
```
